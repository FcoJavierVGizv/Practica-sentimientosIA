```python

from transformers import pipeline

clasificador = pipeline(
    "sentiment-analysis",
    model="pysentimiento/robertuito-sentiment-analysis"
)


lista_negra = [
    "spam", "publicidad", "oferta exclusiva", "gana dinero",
    "dinero fácil", "haz clic", "click aquí", "suscríbete",
    "promoción", "regalo gratis", "premio", "sorteo",

    "estafa", "estafador", "fraude", "timo", "timador",
    "phishing", "robo", "contraseña", "datos bancarios",
    "inversión garantizada", "multiplica tu dinero",

    "idiota", "imbécil", "estúpido", "inútil", "payaso",
    "patético", "basura", "ridículo", "asqueroso",
    "cretino", "gilipollas", "capullo", "cabrón",

    "das pena", "no vales nada", "eres un inútil",
    "eres basura", "cállate", "vete a la mierda",
    "te odio", "qué asco das", "no sirves para nada",
    "menudo inútil", "eres patético",

    "hazte rico", "trabaja desde casa", "dinero rápido",
    "premio asegurado", "reclama tu premio",
    "verifica tu cuenta", "cuenta bloqueada",
    "gana miles", "sin riesgo", "beneficio garantizado",

    "molesto", "insoportable", "horrible", "asco",
    "vergonzoso", "decepcionante", "malísimo",
    "fraudulento", "engaño", "mentira"
]

mensajes = [
    "Me gustan los videojuegos",
    "Este juego es horrible",
    "Esto es spam y no me interesa",
    "Me encanta jugar con mis amigos",
    "La experiencia ha sido mala",
    "Los videojuegos son divertidos",
    "El creador de este videojuego es un completo gilipollas",
    "Este precio es un robo",
    "No me ha gustado nada"
]

mensajes_aceptados = []
mensajes_excluidos = []

for mensaje in mensajes:

    # Comprobamos si contiene alguna palabra de la lista negra
    if any(palabra in mensaje.lower() for palabra in lista_negra):
        mensajes_excluidos.append((mensaje, "Palabra de la lista negra"))
        continue

    # Analizamos el sentimiento del mensaje
    resultado = clasificador(mensaje)[0]

    # Excluimos los mensajes negativos
    if resultado["label"] == "NEG":
        mensajes_excluidos.append((mensaje, "Sentimiento negativo"))
    else:
        mensajes_aceptados.append(mensaje)


print("=== MENSAJES ACEPTADOS ===")
for mensaje in mensajes_aceptados:
    print("-", mensaje)


print("\n=== MENSAJES EXCLUIDOS ===")
for mensaje, motivo in mensajes_excluidos:
    print("-", mensaje, "| Motivo:", motivo)
```
